# Q3 Customer Purchase Analysis

Identify the top 5 customers based on total sales.

Hint: Use GroupBy with aggregation.

```python
#tpo 5 customer sbased o total Sales.

df['Total_sales'] = df['Sales'] * df['Quantity']
print(df)

df1 = df.groupby('Customer')['Total_sales'].sum().nlargest(5)
print('Top 5 Customers', df1)
```

# Q4 Regional Sales Performance

Calculate the total sales generated by each region.

Then identify:

Which region generated the highest revenue?

```python
#Total_Sales generated by each region.
Sales_by_region = df.groupby('Region')['Total_sales'].sum()
print(Sales_by_region)
```

```python
# Find the region with the highest revenue
highest_region = Sales_by_region.idxmax()
highest_revenue = Sales_by_region.max()

print("Region with highest revenue:", highest_region)
print("Highest revenue:", highest_revenue)
```

# Q5 Product Performance Analysis

Calculate the total quantity sold for each product.

Which product appears to be the most popular?



```python
#The Total Quantity for each Product.
Quantity_of_each_products = df.groupby('Product')['Quantity'].sum()
print(Quantity_of_each_products)
```

```python
#Most popular product
most_popular_products = Quantity_of_each_products.idxmax()
print(most_popular_products)
```

# Q7 Pivot Table Creation

Create a pivot table showing:

Rows → Region

Columns → Product

Values → Total Sales

Which region sells the most laptops?

```python
from numpy._core.defchararray import index
pivot = pd.pivot_table(df, values='Total_sales', index='Region', columns='Product', aggfunc='sum')
print(pivot)
```

```python
Sell_most_laptops = pivot.idxmax()  #Which region sells the most laptops?
print(Sell_most_laptops)
```

# Q8 Profitability Analysis

Calculate the average profit for each product category.

Which category is most profitable?

```python
from numpy import average
#Profit for each product category.
average_profit_each_category = df.groupby('Category')['Profit'].mean()
print(average_profit_each_category)
```

```python
#Which category is most profitable?
profitable_category = average_profit_each_category.idxmax()
print(profitable_category)
```
